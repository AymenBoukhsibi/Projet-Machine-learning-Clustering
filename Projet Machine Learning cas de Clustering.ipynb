## Clustering

```python
df.head()
```

### Clustering avec 2 variables

```python

df1 = df[['Annual Income (k$)', 'Spending Score (1-100)']]
df1.head()
```

### Estimation du nombre de classes par la méthode de coude

```python
# scatter plot
sns.scatterplot(x=df1['Annual Income (k$)'], y=df1['Spending Score (1-100)'])
```

```python
from sklearn.cluster import KMeans
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(df1)
    wcss.append(kmeans.inertia_)
```

```python
# Représenation de WCSS en fonction du nopmbre de cluster
plt.figure(figsize=(6,3))
plt.plot(range(1,11), wcss)
plt.plot(range(1,11), wcss, linewidth=3, color='red', marker='8')
plt.xlabel('Nombre de cluster')
plt.ylabel('WCSS')
plt.xticks(np.arange(1,11,1))
plt.show()
```

### Modélisation avec 5 Classes

```python
km = KMeans(n_clusters=5)
km.fit(df1)
```

### Affichages des centres de chaque classes

```python
km.cluster_centers_
```

### Comptage des individus de chaque classe

```python
unique, counts = np.unique(km.labels_, return_counts=True)
print (unique, counts)
```

### Affectation de chaque individue à sa classe

```python
y = km.predict(df1)
df1['Classe'] = y
df1.head()
```

```python
sns.scatterplot(x='Annual Income (k$)', y='Spending Score (1-100)', 
                data=df1, hue='Classe', s=50, palette=['red', 'green', 'brown', 'blue', 'orange'])
```
